```python
# ==========================================================
# Experiment 7: Hash Functions and Avalanche Effect
# Aim:
# Compute MD5, SHA-1 and SHA-256 hashes and demonstrate
# the avalanche effect—a single-character change
# ("hello" → "Hello") produces a completely different digest.
# ==========================================================

import hashlib

# ----------------------------------------------------------
# Step 1: Function to Compute Hashes
# ----------------------------------------------------------
def compute_hashes(data: bytes):
    return {
        "MD5": hashlib.md5(data).hexdigest(),
        "SHA1": hashlib.sha1(data).hexdigest(),
        "SHA256": hashlib.sha256(data).hexdigest()
    }

# ----------------------------------------------------------
# Step 2: Original and Tampered Data
# ----------------------------------------------------------
original_data = b"hello"
tampered_data = b"Hello"

# Compute hashes
hashes_original = compute_hashes(original_data)
hashes_tampered = compute_hashes(tampered_data)

# Repeat hashing to verify consistency
hashes_original_repeat = compute_hashes(original_data)

# ----------------------------------------------------------
# Step 3: Display Results
# ----------------------------------------------------------
print("hello ->", hashes_original)
print("Hello ->", hashes_tampered)

# ----------------------------------------------------------
# Step 4: Test Cases
# ----------------------------------------------------------
def test_experiment7():

    assert hashes_original == hashes_original_repeat, \
        "Same input must always produce the same hash"

    assert hashes_original["MD5"] != hashes_tampered["MD5"], \
        "A single-character change must alter the MD5 hash"

    assert hashes_original["SHA256"] != hashes_tampered["SHA256"], \
        "A single-character change must alter the SHA-256 hash"

    print("Experiment 7: All test cases passed.")

# Run test cases
test_experiment7()
```
