```python
import os

# -------------------------------
# Experiment 8: File Carving (Data Recovery)
# -------------------------------

# Function to create a fake JPEG file
def make_fake_jpeg(payload: bytes):
    return b"\xff\xd8\xff" + payload + b"\xff\xd9"

# Create two fake JPEG files
jpeg1 = make_fake_jpeg(b"PHOTO_OF_SUSPECT_CAR")
jpeg2 = make_fake_jpeg(b"CCTV_FRAME_CAPTURE")

# Simulate a damaged disk by embedding the JPEGs inside random bytes
raw_disk_blob = (
    os.urandom(30) +
    jpeg1 +
    os.urandom(40) +
    jpeg2 +
    os.urandom(20)
)

# File carving function
def carve_jpegs(blob: bytes):
    recovered = []
    start_marker = b"\xff\xd8\xff"
    end_marker = b"\xff\xd9"

    pos = 0

    while True:
        # Find JPEG start signature
        start = blob.find(start_marker, pos)

        if start == -1:
            break

        # Find JPEG end signature
        end = blob.find(end_marker, start)

        if end == -1:
            break

        end += len(end_marker)

        recovered.append(blob[start:end])

        pos = end

    return recovered


# Recover embedded JPEGs
recovered_files = carve_jpegs(raw_disk_blob)

print("Recovered", len(recovered_files), "file(s) via carving")


# -------------------------------
# Test Cases
# -------------------------------

def test_experiment8():
    assert len(recovered_files) == 2, \
        "File carving should recover exactly 2 embedded JPEGs"

    assert recovered_files[0] == jpeg1, \
        "First recovered file must exactly match the original embedded JPEG"

    assert recovered_files[1] == jpeg2, \
        "Second recovered file must exactly match the original embedded JPEG"

    print("Experiment 8: All test cases passed.")


# Run the tests
test_experiment8()
```
