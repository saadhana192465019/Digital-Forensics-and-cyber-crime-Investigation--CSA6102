```python
# ============================================================
# EXPERIMENT 1
# MAPPING A CYBER INCIDENT TO PROVISIONS OF THE IT ACT, 2000
# Rule-Based Legal Classifier
# ============================================================

import re
from dataclasses import dataclass, field

# ============================================================
# 1. IT ACT KNOWLEDGE BASE
# ============================================================

IT_ACT = {

    "43": {
        "title": "Penalty for damage to computer, computer system etc. (civil)",
        "keywords": [
            "unauthorised access",
            "without permission",
            "downloaded copy",
            "introduced virus",
            "damaged",
            "denial of service",
            "disrupted",
            "stole source code"
        ],
        "penalty": "Civil liability - compensation to the affected person"
    },

    "43A": {
        "title": "Compensation for failure to protect sensitive personal data",
        "keywords": [
            "body corporate",
            "negligent",
            "sensitive personal data",
            "failed to protect",
            "reasonable security practices",
            "data leak by company"
        ],
        "penalty": "Compensation to the affected person"
    },

    "65": {
        "title": "Tampering with computer source documents",
        "keywords": [
            "source code",
            "altered source",
            "concealed source",
            "destroyed source document"
        ],
        "penalty": "Up to 3 years imprisonment or fine up to Rs.2,00,000 or both"
    },

    "66": {
        "title": "Computer related offences",
        "keywords": [
            "hacking",
            "dishonestly",
            "fraudulently",
            "unauthorised access",
            "malware",
            "ransomware",
            "deleted data"
        ],
        "penalty": "Up to 3 years imprisonment or fine up to Rs.5,00,000 or both"
    },

    "66B": {
        "title": "Dishonestly receiving stolen computer resource or communication device",
        "keywords": [
            "stolen laptop",
            "stolen mobile",
            "received stolen",
            "knowing it to be stolen"
        ],
        "penalty": "Up to 3 years imprisonment or fine up to Rs.1,00,000 or both"
    },

    "66C": {
        "title": "Identity theft",
        "keywords": [
            "identity theft",
            "used his password",
            "otp",
            "impersonated login",
            "stolen credentials",
            "digital signature misuse"
        ],
        "penalty": "Up to 3 years imprisonment and fine up to Rs.1,00,000"
    },

    "66D": {
        "title": "Cheating by personation by using computer resource",
        "keywords": [
            "phishing",
            "fake website",
            "cheating",
            "posed as bank",
            "vishing",
            "fraudulent call",
            "fake profile to cheat",
            "upi fraud"
        ],
        "penalty": "Up to 3 years imprisonment and fine up to Rs.1,00,000"
    },

    "66E": {
        "title": "Violation of privacy",
        "keywords": [
            "private area",
            "without consent image",
            "hidden camera",
            "obscene photograph of person",
            "voyeurism"
        ],
        "penalty": "Up to 3 years imprisonment or fine up to Rs.2,00,000 or both"
    },

    "66F": {
        "title": "Cyber terrorism",
        "keywords": [
            "cyber terrorism",
            "threaten unity",
            "critical information infrastructure",
            "sovereignty",
            "terror"
        ],
        "penalty": "Imprisonment which may extend to imprisonment for life"
    },

    "67": {
        "title": "Publishing or transmitting obscene material in electronic form",
        "keywords": [
            "obscene",
            "lascivious",
            "published obscene material"
        ],
        "penalty": "First conviction up to 3 years and fine up to Rs.5,00,000"
    },

    "67C": {
        "title": "Preservation and retention of information by intermediaries",
        "keywords": [
            "intermediary",
            "failed to retain",
            "did not preserve logs",
            "retention period"
        ],
        "penalty": "Up to 3 years imprisonment and fine"
    },

    "69": {
        "title": "Power to issue directions for interception, monitoring or decryption",
        "keywords": [
            "interception",
            "monitoring",
            "decryption",
            "government direction",
            "refused to assist agency"
        ],
        "penalty": "Up to 7 years imprisonment and fine for non-compliance"
    },

    "70": {
        "title": "Protected system",
        "keywords": [
            "protected system",
            "critical infrastructure",
            "power grid",
            "notified protected"
        ],
        "penalty": "Up to 10 years imprisonment and fine"
    },

    "72": {
        "title": "Breach of confidentiality and privacy",
        "keywords": [
            "official disclosed",
            "breach of confidentiality",
            "statutory power",
            "register disclosed"
        ],
        "penalty": "Up to 2 years imprisonment or fine up to Rs.1,00,000 or both"
    },

    "72A": {
        "title": "Disclosure of information in breach of lawful contract",
        "keywords": [
            "service provider disclosed",
            "breach of contract",
            "employee leaked customer data",
            "disclosed personal information"
        ],
        "penalty": "Up to 3 years imprisonment or fine up to Rs.5,00,000 or both"
    }
}


# ============================================================
# 2. DATA CLASS FOR MATCHING RESULTS
# ============================================================

@dataclass
class Match:
    section: str
    title: str
    score: int
    matched_terms: list = field(default_factory=list)
    penalty: str = ""


# ============================================================
# 3. OFFENCE MAPPING FUNCTION
# ============================================================

def map_offence(description, kb=IT_ACT, top_n=3):

    text = description.lower()

    results = []

    for sec, meta in kb.items():

        hits = [
            keyword
            for keyword in meta["keywords"]
            if keyword in text
        ]

        if hits:

            results.append(
                Match(
                    section=sec,
                    title=meta["title"],
                    score=len(hits),
                    matched_terms=hits,
                    penalty=meta["penalty"]
                )
            )

    # Highest score first
    results.sort(
        key=lambda m: (-m.score, m.section)
    )

    return results[:top_n]


# ============================================================
# 4. REPORT FUNCTION
# ============================================================

def report(description):

    matches = map_offence(description)

    lines = []

    lines.append("INCIDENT: " + description)
    lines.append("-" * 70)

    if not matches:
        lines.append(
            "No section matched. Refer the matter to a legal expert."
        )

    else:

        for m in matches:

            lines.append(
                f"Section {m.section}: {m.title}"
            )

            lines.append(
                f"Matched terms: {', '.join(m.matched_terms)}"
            )

            lines.append(
                f"Score: {m.score}"
            )

            lines.append(
                f"Indicative penalty: {m.penalty}"
            )

            lines.append("")

    lines.append(
        "NOTE: Educational mapping only - not legal advice."
    )

    return "\n".join(lines)


# ============================================================
# 5. TEST CASES
# ============================================================

def run_tests():

    cases = [

        (
            "An attacker created a fake website of a bank and "
            "used phishing emails to cheat customers.",
            "66D"
        ),

        (
            "An employee used his colleague's password and OTP "
            "to commit identity theft on the payroll portal.",
            "66C"
        ),

        (
            "A programmer deliberately altered the source code "
            "of the billing software and concealed it.",
            "65"
        ),

        (
            "A group targeted the power grid, a notified protected "
            "system, to threaten the sovereignty of the nation.",
            "66F"
        ),

        (
            "A body corporate was negligent in handling sensitive "
            "personal data and failed to protect it.",
            "43A"
        ),

        (
            "A call centre employee leaked customer data in breach "
            "of contract and disclosed personal information.",
            "72A"
        ),

        (
            "Ransomware was used to dishonestly encrypt and delete "
            "data after unauthorised access.",
            "66"
        ),

        (
            "Someone installed a hidden camera and captured the "
            "private area of a person without consent.",
            "66E"
        )
    ]

    passed = 0

    print("=" * 75)
    print("IT ACT 2000 - TEST CASE RESULTS")
    print("=" * 75)

    for i, (description, expected) in enumerate(cases, 1):

        matches = map_offence(description)

        got = [
            m.section
            for m in matches
        ]

        ok = expected in got

        if ok:
            passed += 1

        status = "PASS" if ok else "FAIL"

        print(
            f"TC{i:02d} | Expected: s.{expected:<4} "
            f"| Got: {got} | {status}"
        )

    # --------------------------------------------------------
    # Negative Test
    # --------------------------------------------------------

    negative_description = (
        "The employee submitted a leave application "
        "in the HR portal."
    )

    negative_result = map_offence(
        negative_description
    )

    negative_pass = len(negative_result) == 0

    if negative_pass:
        passed += 1

    print(
        f"TC09 | Negative case | "
        f"No section returned | "
        f"{'PASS' if negative_pass else 'FAIL'}"
    )

    print("-" * 75)

    total = len(cases) + 1

    print(
        f"RESULT: {passed}/{total} test cases passed"
    )

    print("=" * 75)

    return passed == total


# ============================================================
# 6. RUN TESTS
# ============================================================

run_tests()


# ============================================================
# 7. SAMPLE INCIDENT
# ============================================================

sample_incident = (
    "An attacker created a fake website of a bank "
    "and used phishing emails to cheat customers."
)

print("\n")
print("=" * 75)
print("SAMPLE INCIDENT ANALYSIS")
print("=" * 75)

print(
    report(sample_incident)
)


# ============================================================
# 8. USER INPUT - ENTER YOUR OWN CYBER INCIDENT
# ============================================================

print("\n")
print("=" * 75)
print("CUSTOM CYBER INCIDENT ANALYSIS")
print("=" * 75)

incident = input(
    "Enter your cyber incident description: "
)

print("\n")
print(report(incident))
```
