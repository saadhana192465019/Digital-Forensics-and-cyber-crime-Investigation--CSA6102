```python
# ==========================================================
# Experiment 5: Forensic Imaging vs Ordinary Duplication
# Aim:
# Show that forensic imaging captures a storage device
# sector-by-sector (including deleted data), while ordinary
# duplication only copies the active, visible regions.
# ==========================================================

import os

# ----------------------------------------------------------
# Step 1: Create Sample Storage Device
# ----------------------------------------------------------
storage_file = "storage_device.bin"

# Active data (9 bytes)
active_data = b"ACTIVE123"

# Deleted/unused data (63 bytes)
deleted_data = b"\x00" * 63

# Total image = 72 bytes
with open(storage_file, "wb") as f:
    f.write(active_data)
    f.write(deleted_data)

# ----------------------------------------------------------
# Step 2: Forensic Imaging
# Copies the entire storage device
# ----------------------------------------------------------
def forensic_imaging(filename):
    with open(filename, "rb") as f:
        return f.read()

# ----------------------------------------------------------
# Step 3: Ordinary Duplication
# Copies only the active (visible) data
# ----------------------------------------------------------
def ordinary_duplication(filename):
    with open(filename, "rb") as f:
        return f.read(9)

# ----------------------------------------------------------
# Perform Imaging
# ----------------------------------------------------------
forensic_image = forensic_imaging(storage_file)
duplicate = ordinary_duplication(storage_file)

# ----------------------------------------------------------
# Display Output
# ----------------------------------------------------------
print("Forensic image size:", len(forensic_image))
print("Duplication size:", len(duplicate))

# ----------------------------------------------------------
# Test Cases
# ----------------------------------------------------------
def test_experiment5():
    assert len(forensic_image) == 72
    assert len(duplicate) == 9
    assert len(forensic_image) > len(duplicate)
    print("\nExperiment 5: All test cases passed.")

test_experiment5()
```
