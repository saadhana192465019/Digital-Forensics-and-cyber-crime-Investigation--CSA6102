```python
# ==========================================================
# Experiment 3: Data Acquisition Types
# Aim:
# Demonstrate Physical, Logical and Sparse Data Acquisition.
# ==========================================================

import os

# ----------------------------------------------------------
# Step 1: Create Sample Evidence Folder
# ----------------------------------------------------------
evidence_dir = "acquisition_evidence"
os.makedirs(evidence_dir, exist_ok=True)

# Create sample files
sample_files = {
    "photo.jpg": "This is a sample photo file.",
    "report.docx": "This is a confidential report."
}

for filename, content in sample_files.items():
    with open(os.path.join(evidence_dir, filename), "w") as f:
        f.write(content)

# ----------------------------------------------------------
# Step 2: Physical Acquisition
# (Includes all file data + deleted/unused bytes simulation)
# ----------------------------------------------------------
def physical_acquisition(folder):
    image = b""

    for filename in sorted(os.listdir(folder)):
        with open(os.path.join(folder, filename), "rb") as f:
            image += f.read()

    # Simulate deleted/unused space
    image += b"\x00" * 10

    return image

# ----------------------------------------------------------
# Step 3: Logical Acquisition
# (Only visible files)
# ----------------------------------------------------------
def logical_acquisition(folder):
    return sorted(os.listdir(folder))

# ----------------------------------------------------------
# Step 4: Sparse Acquisition
# (Only selected files)
# ----------------------------------------------------------
def sparse_acquisition(folder, selected_files):
    return [f for f in sorted(os.listdir(folder)) if f in selected_files]

# ----------------------------------------------------------
# Perform Acquisition
# ----------------------------------------------------------
physical_image = physical_acquisition(evidence_dir)
logical_image = logical_acquisition(evidence_dir)
sparse_image = sparse_acquisition(evidence_dir, ["report.docx"])

# ----------------------------------------------------------
# Print Output
# ----------------------------------------------------------
print("Physical image size:", len(physical_image), "bytes")
print("Logical image files:", logical_image)
print("Sparse image files:", sparse_image)

# ----------------------------------------------------------
# Test Cases
# ----------------------------------------------------------
def test_experiment3():
    assert len(physical_image) > 0
    assert logical_image == ["photo.jpg", "report.docx"]
    assert sparse_image == ["report.docx"]
    assert len(physical_image) > sum(
        os.path.getsize(os.path.join(evidence_dir, f))
        for f in logical_image
    )
    print("Experiment 3: All test cases passed.")

test_experiment3()
```
