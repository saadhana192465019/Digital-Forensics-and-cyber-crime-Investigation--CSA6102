```python
# ==========================================================
# Experiment 4: Live Acquisition vs Dead Acquisition
# Aim:
# Show that live acquisition (system ON) captures volatile
# data, while dead acquisition (system OFF) captures only
# non-volatile data.
# ==========================================================

import os
import time
import sys
import subprocess

# Install psutil if not already installed
try:
    import psutil
except ImportError:
    subprocess.check_call([sys.executable, "-m", "pip", "install", "psutil"])
    import psutil

# ----------------------------------------------------------
# Create Sample Hard Disk Files
# ----------------------------------------------------------
disk_folder = "dead_acquisition"

os.makedirs(disk_folder, exist_ok=True)

sample_files = {
    "a.txt": "This is file A.",
    "b.txt": "This is file B."
}

for filename, content in sample_files.items():
    with open(os.path.join(disk_folder, filename), "w") as f:
        f.write(content)

# ----------------------------------------------------------
# Live Acquisition (System ON)
# ----------------------------------------------------------
def live_acquisition():
    processes = [p.info for p in psutil.process_iter(['pid'])]

    return {
        "timestamp": time.time(),
        "running_processes": len(processes),
        "cpu_percent": psutil.cpu_percent(interval=0.1)
    }

# ----------------------------------------------------------
# Dead Acquisition (System OFF)
# ----------------------------------------------------------
def dead_acquisition(folder):
    return {
        "hard_disk_files": sorted(os.listdir(folder)),
        "ram_data": None
    }

# ----------------------------------------------------------
# Capture Data
# ----------------------------------------------------------
live_data = live_acquisition()
dead_data = dead_acquisition(disk_folder)

# ----------------------------------------------------------
# Display Output
# ----------------------------------------------------------
print("Live acquisition:", live_data)
print("Dead acquisition:", dead_data)

# ----------------------------------------------------------
# Test Cases
# ----------------------------------------------------------
def test_experiment4():
    assert live_data["running_processes"] > 0
    assert isinstance(live_data["timestamp"], float)
    assert dead_data["hard_disk_files"] == ["a.txt", "b.txt"]
    assert dead_data["ram_data"] is None
    print("Experiment 4: All test cases passed.")

test_experiment4()
```
