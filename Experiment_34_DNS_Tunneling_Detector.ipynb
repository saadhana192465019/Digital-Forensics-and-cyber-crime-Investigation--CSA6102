```python
# ==========================================================
# Experiment 6: DNS Tunneling Detector
# ==========================================================

import math
from collections import Counter


# ----------------------------------------------------------
# Function to calculate Shannon Entropy
# ----------------------------------------------------------
def shannon_entropy(s):
    """
    Calculate the Shannon entropy of a string.
    Higher entropy indicates more randomness.
    """

    if not s:
        return 0.0

    counts = Counter(s)
    length = len(s)

    entropy = -sum(
        (count / length) * math.log2(count / length)
        for count in counts.values()
    )

    return entropy


# ----------------------------------------------------------
# Function to detect possible DNS tunneling
# ----------------------------------------------------------
def detect_dns_tunneling(
    queries,
    length_threshold=20,
    entropy_threshold=3.5
):
    """
    Detect DNS queries with long, high-entropy
    subdomain labels.
    """

    flagged = []

    for query in queries:

        # Extract leftmost label
        label = query.split(".")[0]

        entropy = shannon_entropy(label)

        if (
            len(label) >= length_threshold
            and entropy >= entropy_threshold
        ):

            flagged.append({
                "query": query,
                "label_length": len(label),
                "entropy": round(entropy, 2)
            })

    return flagged


# ==========================================================
# Test Cases
# ==========================================================

def test_experiment6():

    queries = [

        "www.google.com",

        "mail.office365.com",

        "a8f3k2j9x1p7q4z6w0n5r2t8y3.exfil-domain.com",

        "vpn.corporate-network.com",

        "9c2e7b1a4f8d3c6e0a5b9d2f7c1e4a8b3d6f9c2e5a8b1d4f.tunnel.example.net"

    ]

    # Detect suspicious DNS queries
    flagged = detect_dns_tunneling(queries)

    # Display Results
    print("DNS Tunneling Detection Results")
    print("-" * 65)

    if flagged:

        for item in flagged:
            print(f"Query          : {item['query']}")
            print(f"Label Length   : {item['label_length']}")
            print(f"Entropy        : {item['entropy']}")
            print()

    else:
        print("No suspicious DNS queries detected.")

    # Assertions
    flagged_domains = [item["query"] for item in flagged]

    assert "www.google.com" not in flagged_domains
    assert "mail.office365.com" not in flagged_domains
    assert "vpn.corporate-network.com" not in flagged_domains

    assert "a8f3k2j9x1p7q4z6w0n5r2t8y3.exfil-domain.com" in flagged_domains

    assert (
        "9c2e7b1a4f8d3c6e0a5b9d2f7c1e4a8b3d6f9c2e5a8b1d4f.tunnel.example.net"
        in flagged_domains
    )

    assert len(flagged) == 2

    print("All test cases passed.")


# ----------------------------------------------------------
# Run Test
# ----------------------------------------------------------

test_experiment6()
```
